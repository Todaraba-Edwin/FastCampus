### 첫째, 클래스와 객체지향 프로그래밍
#### (1) 클래스 정의 및 기본 개념

- 클래스(class): 객체를 만드는 청사진
- 인스턴스(instance): 클래스로부터 생성된 개별 객체
- `__init__()`: 생성자, 인스턴스 초기화 메서드
- `self`: 인스턴스 자신을 참조하는 매개변수

````python
class Car:
    # 클래스 변수
    wheel = 4
    
    # 생성자
    def __init__(self, model, color):
        self.model = model
        self.color = color
    
    # 메서드
    def drive(self):
        return f"{self.model}이 {self.color}색으로 달린다"

# 인스턴스 생성
car1 = Car("Tesla", "빨강")
print(car1.drive())
print(f"바퀴 개수: {car1.wheel}")
````

#### (2) 상속과 다형성

- 상속(inheritance): 기존 클래스의 속성과 메서드를 새로운 클래스가 물려받음
- 부모 클래스: 상속을 해주는 클래스
- 자식 클래스: 상속을 받는 클래스
- 오버라이딩(overriding): 부모 클래스의 메서드를 자식 클래스에서 재정의

````python
# 부모 클래스
class Vehicle:
    def __init__(self, name):
        self.name = name
    
    def move(self):
        return f"{self.name}이 움직인다"

# 자식 클래스
class Bicycle(Vehicle):
    def move(self):
        return f"{self.name}이 자전거로 움직인다"

class Airplane(Vehicle):
    def move(self):
        return f"{self.name}이 하늘을 난다"

# 다형성 확인
vehicles = [Bicycle("자전거"), Airplane("비행기")]
for vehicle in vehicles:
    print(vehicle.move())
````

### 둘째, 모듈과 패키지
#### (1) 모듈 작성 및 사용

- 모듈(module): 함수, 클래스, 변수를 정의한 .py 파일
- `import`: 모듈 전체 임포트
- `from ... import`: 모듈의 특정 항목 임포트
- 모듈의 `__name__` == `"__main__"`: 모듈이 직접 실행되는 경우

````python
# 표준 라이브러리 모듈 임포트
import math
from datetime import datetime

# 원의 넓이 계산
radius = 5
area = math.pi * radius ** 2
print(f"원의 넓이: {area:.2f}")

# 현재 시간
now = datetime.now()
print(f"현재 시간: {now}")
````

#### (2) 패키지와 __init__.py

- 패키지(package): 모듈을 포함하는 디렉토리
- `__init__.py`: 디렉토리를 패키지로 인식하게 하는 파일
- 패키지 구조: 폴더 내 모듈들을 계층적으로 구성
- 절대 임포트 vs 상대 임포트

```
mypackage/
├── __init__.py
├── module1.py
├── module2.py
└── subpackage/
    ├── __init__.py
    └── module3.py
```

```python
# 절대 임포트
from mypackage.module1 import function1
from mypackage.subpackage.module3 import function3

# 상대 임포트
from .module2 import function2
from ..module1 import function1
```

### 셋째, 예외처리
#### (1) 예외 처리의 기본

- 예외(exception): 프로그램 실행 중 발생하는 오류
- try-except 블록: 예외 발생 시 처리
- else: 예외가 발생하지 않았을 때 실행
- finally: 예외 발생 여부와 관계없이 항상 실행

````python
try:
    number = int(input("숫자를 입력: "))
    result = 10 / number
except ValueError:
    print("숫자가 아닙니다")
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다")
except Exception as e:
    print(f"예상치 못한 오류: {e}")
else:
    print(f"계산 결과: {result}")
finally:
    print("프로그램 종료")
````

#### (2) 예외 발생 및 처리

- `raise`: 의도적으로 예외 발생
- 사용자 정의 예외: 프로젝트에 맞는 예외 클래스 작성
- `except ... as e`: 예외 객체 정보 접근
- 예외 체이닝: 원인 예외를 포함한 새로운 예외 발생

````python
# 사용자 정의 예외
class NegativeNumberError(Exception):
    pass

def check_age(age):
    if age < 0:
        raise NegativeNumberError("나이는 음수가 될 수 없습니다")
    return f"{age}살입니다"

try:
    print(check_age(25))
    print(check_age(-5))
except NegativeNumberError as e:
    print(f"오류: {e}")
````

### 넷째, 내장함수와 라이브러리
#### (1) 내장함수 (map, filter, lambda)

- `map()`: 각 요소에 함수 적용
- `filter()`: 조건을 만족하는 요소만 필터링
- `lambda`: 간단한 익명 함수
- `reduce()`: 누적 계산 (functools 모듈)

````python
# map 사용
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
print(f"제곱: {squared}")

# filter 사용
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(f"짝수: {even_numbers}")

# reduce 사용
from functools import reduce
total = reduce(lambda x, y: x + y, numbers)
print(f"합계: {total}")

for i, name in enumerate(['body', 'foo', 'bar']):
    print(f'{i}_{name}', end=' ')
````

#### (2) 표준 라이브러리

- `os`: 운영체제 상호작용 (파일, 디렉토리)
- `sys`: Python 인터프리터 관련 (명령행 인자)
- `datetime`: 날짜와 시간 처리
- `json`: JSON 파일 읽기/쓰기
- `random`: 난수 생성

````python
import os
import sys
from datetime import datetime, timedelta
import json
import random

# os: 현재 디렉토리
print(f"현재 경로: {os.getcwd()}")

# datetime: 날짜 계산
today = datetime.now()
tomorrow = today + timedelta(days=1)
print(f"내일: {tomorrow.date()}")

# json: 딕셔너리를 JSON 문자열로
data = {"name": "Alice", "age": 30}
json_str = json.dumps(data)
print(f"JSON: {json_str}")

# random: 랜덤 선택
choices = random.sample(range(1, 46), 6)
print(f"로또 번호: {sorted(choices)}")
````

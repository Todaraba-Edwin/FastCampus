## 첫째, 함수
### (1) 함수 정의와 호출

#### 구조

```python
def 함수명():
    실행할 코드

함수명()  # 호출
```

**JavaScript와의 차이:**
- JS: `function name() {}` → Python: `def name():`
- JS: 중괄호 `{}` → Python: **들여쓰기**
- 함수 호출은 동일: `name()`

````python
# 가장 간단한 함수
def greet():
    print("안녕하세요!")

greet()  # 함수 호출
````

````python
# 함수를 여러 번 호출
def say_hello():
    print("Hello!")

say_hello()
say_hello()
say_hello()
````

````python
# 계산을 수행하는 함수
def calculate():
    result = 10 + 20
    print(f"계산 결과: {result}")

calculate()
````

### (2) 매개변수와 반환값

#### 구조

```python
def 함수명(매개변수1, 매개변수2):
    처리 코드
    return 반환값

결과 = 함수명(인자1, 인자2)
```

**특징:**
- **매개변수**: 함수 정의 시 괄호 안의 변수
- **인자**: 함수 호출 시 전달하는 값
- **return**: 함수의 결과값을 반환

````python
# 매개변수가 있는 함수
def greet_person(name):
    print(f"안녕하세요, {name}님!")

greet_person("김철수")
greet_person("이영희")
````

````python
# 여러 매개변수
def add(a, b):
    result = a + b
    print(f"{a} + {b} = {result}")

add(10, 20)
add(100, 50)
````

````python
# return으로 값 반환
def multiply(x, y):
    return x * y  # ⭐ 결과값을 반환

result1 = multiply(3, 4)
result2 = multiply(5, 6)

print(f"3 × 4 = {result1}")
print(f"5 × 6 = {result2}")
````

````python
# 여러 값을 반환
def get_min_max(numbers):
    return min(numbers), max(numbers)  # 튜플로 반환

data = [3, 1, 4, 1, 5, 9, 2, 6]
min_val, max_val = get_min_max(data)  # ⭐ 여러 값을 한 번에 받기

print(f"최솟값: {min_val}, 최댓값: {max_val}")
````

````python
# 기본값이 있는 매개변수
def greet_with_message(name, message="좋은 하루 되세요!"):
    print(f"{name}님, {message}")

greet_with_message("김철수")  # message는 기본값 사용
greet_with_message("이영희", "반가워요!")  # message를 지정
````

### (3) 기본값 매개변수 위치 규칙

#### 주의사항: 기본값 매개변수 위치

**기본값이 있는 매개변수는 뒤에 와야 함** ⭐

```python
# ❌ 틀림: 기본값 매개변수가 앞에 옴
def greet(name="김철수", age):  # SyntaxError!
    print(f"{name}님, {age}살입니다")

# ✅ 맞음: 기본값 없는 매개변수가 앞, 기본값 있는 매개변수가 뒤
def greet(age, name="김철수"):
    print(f"{name}님, {age}살입니다")
```

**JavaScript와의 차이:**
- JS: 기본값 순서 제약 없음
- Python: 기본값이 있는 매개변수는 항상 뒤에 와야 함

````python
# 올바른 기본값 매개변수 위치
def introduce(age, name="김철수", city="서울"):
    print(f"{name}님, {age}살, {city} 거주")

introduce(25)  # 기본값 사용
introduce(30, "이영희")  # name만 지정
introduce(28, "박준호", "부산")  # 모두 지정
````

### (4) 키워드 매개변수 (Keyword Arguments)

#### 개념

**키워드 매개변수**: 매개변수 이름을 명시해서 전달
- 순서와 상관없이 원하는 값 전달 가능
- 가독성 좋음, 실수 줄임

```python
def func(a, b, c):
    return a + b + c

# 위치 기반 (positional arguments)
func(1, 2, 3)

# 키워드 기반 (keyword arguments)
func(c=3, a=1, b=2)  # ⭐ 순서 상관없음
```

````python
# 키워드 매개변수 예제
def create_profile(name, age, job="개발자"):
    print(f"이름: {name}, 나이: {age}, 직업: {job}")

# 위치 기반
create_profile("김철수", 25)

# 키워드 기반 (순서 상관없음)
create_profile(age=30, name="이영희", job="디자이너")
create_profile(job="매니저", name="박준호", age=28)
````

### (5) 가변 매개변수 (*args, **kwargs)

#### *args (Variable Length Arguments)

**개념**
- `*args`: 정해지지 않은 개수의 위치 인자 받기
- 튜플로 변환되어 전달됨
- 이름은 `args`가 아니어도 됨 (`*numbers`, `*items` 등 가능하지만 관례상 `*args` 사용)

```python
def sum_all(*args):  # ⭐ 개수 상관없이 받음
    print(f"받은 인자: {args}")  # 튜플
    return sum(args)

print(sum_all(1, 2, 3))
print(sum_all(10, 20, 30, 40, 50))
```

````python
# *args 예제
def print_items(*args):
    for i, item in enumerate(args, 1):
        print(f"{i}. {item}")

print_items("사과")
print()
print_items("사과", "바나나", "딸기", "포도")
````

#### **kwargs (Keyword Arguments Dictionary)

**개념**
- `**kwargs`: 정해지지 않은 개수의 키워드 인자 받기
- 딕셔너리로 변환되어 전달됨
- 관례상 `**kwargs` 사용

```python
def print_info(**kwargs):  # ⭐ 키워드 인자를 딕셔너리로
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="김철수", age=25)
print_info(name="이영희", age=30, job="개발자", city="서울")
```

````python
# **kwargs 예제
def create_user(**kwargs):
    print(f"사용자 정보: {kwargs}")
    print(f"이름: {kwargs.get('name', '미입력')}")
    print(f"나이: {kwargs.get('age', '미입력')}")

create_user(name="김철수", age=25)
print()
create_user(name="이영희", age=30, job="개발자")
````

#### *args와 **kwargs 함께 사용

**순서 규칙** ⭐
```python
def func(positional, *args, **kwargs):
    # 순서: 일반 매개변수 → *args → **kwargs (필수!)
    pass
```

**이건 규칙이지, 가독성이 아닙니다!**

Python이 함수 호출 시 인자를 분류하기 위한 **강제 규칙**:

```python
# 호출 예
func('value', 1, 2, 3, key1='a', key2='b')
```

파이썬이 이렇게 분석합니다:
- `positional` ← `'value'` (첫 번째 위치 인자)
- `*args` ← `(1, 2, 3)` (남은 위치 인자들)
- `**kwargs` ← `{'key1': 'a', 'key2': 'b'}` (키워드 인자들)

**반대 순서면 안 되는 이유:**

```python
# ❌ 이건 SyntaxError!
def func(positional, **kwargs, *args):
    pass

# 왜? 호출할 때 1, 2, 3을 어디에 할당할지 불명확
# **kwargs가 먼저 오면 모든 키워드를 받으려고 하는데
# 위치 인자 1, 2, 3은 어디 가나? 🤔
```

**정리: 매개변수 순서 (필수)**
1. 일반 위치 매개변수 (name, age 등)
2. *args (가변 위치 인자) → 튜플
3. **kwargs (가변 키워드 인자) → 딕셔너리

````python
# *args와 **kwargs 함께 사용
def flexible_function(name, *scores, **info):
    print(f"이름: {name}")
    print(f"점수들: {scores}")  # 튜플
    print(f"추가 정보: {info}")  # 딕셔너리

flexible_function("김철수", 85, 90, 88)
print()
flexible_function("이영희", 95, 92, job="개발자", department="AI팀")
````

## 둘째, 사용자 입출력
### (1) input() 함수

#### 구조

```python
사용자_입력 = input("프롬프트 메시지")
```

**특징:**
- 사용자로부터 **문자열로** 입력받음 (항상 str 타입)
- 숫자로 사용하려면 **명시적 변환** 필요: `int()`, `float()`
- JavaScript: `prompt()` 와 유사하지만, 기능이 더 단순

````python
# 문자열 입력
name = input("이름을 입력하세요: ")
print(f"입력받은 이름: {name}")
print(f"타입: {type(name)}  # <class 'str'>")
````

````python
# 숫자 입력 (명시적 변환 필수)
age_str = input("나이를 입력하세요: ")
age = int(age_str)  # ⭐ 문자열 → 정수로 변환

print(f"입력받은 나이: {age}")
print(f"타입: {type(age)}  # <class 'int'>")
print(f"10년 후 나이: {age + 10}")
````

````python
# 여러 입력값 처리
num1 = int(input("첫 번째 숫자: "))
num2 = int(input("두 번째 숫자: "))

print(f"{num1} + {num2} = {num1 + num2}")
print(f"{num1} - {num2} = {num1 - num2}")
print(f"{num1} * {num2} = {num1 * num2}")
````

````python
# 실수 입력
price = float(input("상품 가격을 입력하세요: "))
quantity = int(input("수량을 입력하세요: "))

total = price * quantity
print(f"총액: {total:,.0f}원")
````

### (2) print() 심화

#### 기본 사용

```python
print(값1, 값2, 값3)  # 쉼표로 여러 값 출력
```

````python
# end 매개변수 - 끝 문자 변경
print("로딩", end="")
print(".", end="")
print(".", end="")
print(".")  # 마지막에 줄바꿈

print()

# for와 함께 사용
for i in range(1, 6):
    print(i, end=" ")  # 공백으로 끝내기
print()  # 줄바꿈
````

````python
# 숫자 포맷팅
price = 10000
quantity = 5

print(f"상품 가격: {price:,}원")  # 천 단위 구분
print(f"총액: {price * quantity:,}원")
print(f"이진수: {15:b}")  # 이진수
print(f"16진수: {255:x}")  # 16진수
````

## 셋째, 파일 입출력
### (1) 파일 읽기

#### 구조

```python
with open("파일명", "r") as f:
    데이터 = f.read()  # 전체 내용 읽기
```

**특징:**
- `with` 문: 파일 자동 종료 (권장)
- `"r"`: 읽기 모드 (read)
- `read()`: 전체 파일을 문자열로 읽음
- `readlines()`: 줄 단위로 리스트로 읽음
- `for 줄 in f`: 한 줄씩 순회

````python
# 예제용 파일 생성
with open("example.txt", "w") as f:
    f.write("Python은 쉽습니다.\n")
    f.write("프로그래밍을 배워봅시다.\n")
    f.write("파이썬 최고!")

print("파일이 생성되었습니다.")
````

````python
# 전체 파일 내용 읽기
with open("example.txt", "r") as f:
    content = f.read()  # 전체 내용을 문자열로

print("===== 파일 내용 =====")
print(content)
````

````python
# 줄 단위로 읽기 (readlines)
with open("example.txt", "r") as f:
    lines = f.readlines()  # 리스트로 반환

print(f"총 {len(lines)}줄입니다.")
for i, line in enumerate(lines, 1):
    print(f"{i}번째: {line.strip()}")
````

````python
# 한 줄씩 순회 (가장 메모리 효율적)
with open("example.txt", "r") as f:
    for line in f:  # ⭐ 자동으로 한 줄씩
        print(f"[읽음] {line.strip()}")
````

### (2) 파일 쓰기

#### 구조

```python
with open("파일명", "w") as f:
    f.write("내용")  # 파일에 쓰기
```

**특징:**
- `"w"`: 쓰기 모드 (기존 파일 덮어쓰기)
- `"a"`: 추가 모드 (기존 내용 뒤에 추가)
- `write()`: 문자열 쓰기
- `writelines()`: 리스트의 문자열들을 한 번에 쓰기

````python
# 파일에 쓰기 (새로 생성 또는 덮어쓰기)
with open("output.txt", "w") as f:
    f.write("첫 번째 줄\n")
    f.write("두 번째 줄\n")
    f.write("세 번째 줄")

print("output.txt에 내용을 썼습니다.")
````

````python
# 기존 파일에 내용 추가 ("a" 모드)
with open("output.txt", "a", encoding="utf-8") as f:  # ⭐ append 모드
    f.write("\n네 번째 줄 (추가)")

print("파일에 내용을 추가했습니다.")

# 확인
with open("output.txt", "r") as f:
    print("\n===== 파일 내용 =====")
    print(f.read())
````

````python
read = open("output.txt", "r")
````

````python
while True:
    line = read.readline()
    if not line:
        break
    print(line.strip())
   
````

````python
read.close()  # 파일 닫기
````

````python
# 리스트 내용을 파일에 쓰기
data = ["사과", "바나나", "딸기", "포도"]

with open("fruits.txt", "w") as f:
    for fruit in data:
        f.write(fruit + "\n")

print("fruits.txt를 생성했습니다.")
````

````python
# 파일 읽고 처리해서 새 파일에 쓰기
# 입력 파일에서 읽기
with open("fruits.txt", "r") as input_file:
    content = input_file.read()

# 처리: 대문자로 변환
processed = content.upper()

# 처리된 내용을 새 파일에 쓰기
with open("ㄴㅐ.txt", "w") as output_file:
    output_file.write(processed)

print("fruits_upper.txt를 생성했습니다.")

# 확인
with open("fruits_upper.txt", "r") as f:
    print("\n===== 변환된 파일 내용 =====")
    print(f.read())
````

````python
# 정리: 생성한 파일 삭제
import os

files_to_remove = ["example.txt", "output.txt", "fruits.txt", "fruits_upper.txt"]
for file in files_to_remove:
    if os.path.exists(file):
        os.remove(file)
        print(f"{file} 삭제됨")
````

## 넷째, 프로그램의 입출력
### (1) sys.argv 명령행 인자

#### 개념

**sys.argv**: 명령행에서 전달된 인자 받기
- 리스트 형태로 반환됨
- `argv[0]`: 프로그램 파일명
- `argv[1]`, `argv[2]` ... : 전달된 인자들

```python
import sys

# 명령어: python script.py apple banana cherry
# sys.argv = ['script.py', 'apple', 'banana', 'cherry']

print(sys.argv)  # 전체 인자 출력
print(sys.argv[0])  # 프로그램명
print(sys.argv[1:])  # 전달된 인자들만
```

### (2) sys.stdin, sys.stdout, sys.stderr

#### 개념

**표준 입출력 스트림**
- **sys.stdin**: 표준 입력 (기본: 키보드)
- **sys.stdout**: 표준 출력 (기본: 터미널, print() 사용)
- **sys.stderr**: 표준 에러 (기본: 터미널, 오류 메시지)

```python
import sys

# print() 대신 sys.stdout.write() 사용
sys.stdout.write("일반 메시지\n")

# 에러 메시지를 sys.stderr로 출력
sys.stderr.write("에러 발생!\n")

# sys.stdin으로 입력받기 (input()과 유사)
line = sys.stdin.readline()  # 한 줄 읽기 (개행 포함)
```

**input() vs sys.stdin.readline()**
- `input()`: 개행 제거 후 반환
- `sys.stdin.readline()`: 개행 포함해서 반환

````python
import sys

# sys.stdout.write() - print()와 유사하지만 자동 줄바꿈 없음
sys.stdout.write("표준 출력 1")
sys.stdout.write("\n")
sys.stdout.write("표준 출력 2\n")

print()

# sys.stderr.write() - 에러 메시지 출력
sys.stderr.write("⚠️ 경고: 파일을 찾을 수 없습니다!\n")

# 다양한 상황에서 사용
try:
    num = int("abc")
except ValueError:
    sys.stderr.write("❌ 에러: 숫자 변환 실패\n")
````

### (3) sys.exit() 프로그램 종료

#### 개념

**sys.exit()**: 프로그램 즉시 종료
- 종료 상태 코드 지정 가능
- **0**: 정상 종료
- **0 이외의 숫자**: 에러 발생 (관례)

```python
import sys

# 프로그램 정상 종료
sys.exit()  # 또는 sys.exit(0)

# 에러와 함께 종료
sys.exit(1)  # 에러 코드 1

# 메시지와 함께 종료
sys.exit("필수 인자가 없습니다!")  # 메시지 출력 후 종료 (상태: 1)
```

**상태 코드 (exit code)**
- 셀(Shell)에서 마지막 명령어의 상태코드는 `$?` 변수에 저장됨
- 0 = 성공, 0 이외 = 실패

````python
import sys

# 예제 1: 조건에 따라 프로그램 종료
def check_age(age):
    if age < 0:
        sys.exit("❌ 에러: 나이는 음수일 수 없습니다!")
    elif age < 18:
        print("미성년자")
    else:
        print("성인")

# 정상 실행
check_age(25)
print("프로그램 계속 실행됨")

print()

# 예제 2: 파일 처리 (파일 없을 때 종료)
import os

filename = "nonexistent.txt"
if not os.path.exists(filename):
    sys.stderr.write(f"에러: {filename} 파일을 찾을 수 없습니다\n")
    sys.exit(1)  # ⭐ 에러 코드 1로 종료
else:
    print(f"{filename} 파일 처리 중...")
    # 파일 처리
    sys.exit(0)  # ⭐ 정상 종료
````
